 Econtrar la media armonica Encontrar la frecuencia

```python
from functools import reduce


def media_armonica(numeros):
  """Calcula la media armonica de una lista de numeros."""
  suma_inversos = reduce(lambda acum, n: acum + (1 / n), numeros, 0)
  return len(numeros) / suma_inversos


def frecuencia(elementos):
  """Cuenta cuantas veces aparece cada elemento en una lista."""
  conteo = {}
  for elem in elementos:
    conteo[elem] = conteo.get(elem, 0) + 1
  return conteo


# ----------------------------------------------------------------------
# Funciones del menu
# ----------------------------------------------------------------------

def opcion_media_armonica():
  entrada = input("Ingrese numeros separados por coma (ej: 2,4,6): ").strip()
  numeros = [float(n) for n in entrada.split(",")]
  resultado = media_armonica(numeros)
  print(f"\nLa media armonica es: {round(resultado, 4)}")


def opcion_frecuencia():
  entrada = input("Ingrese elementos separados por coma (ej: a,b,a,c,b,a): ").strip()
  elementos = [e.strip() for e in entrada.split(",")]
  resultado = frecuencia(elementos)
  print("\nFrecuencia de cada elemento:")
  for elem, veces in resultado.items():
    print(f"  {elem}: {veces}")


# ----------------------------------------------------------------------
# Menu principal
# ----------------------------------------------------------------------

def mostrar_menu():
  print("\n===== MENU =====")
  print("1. Calcular media armonica")
  print("2. Calcular frecuencia de elementos")
  print("0. Salir")


def main():
  opciones = {
      "1": opcion_media_armonica,
      "2": opcion_frecuencia,
  }

  while True:
    mostrar_menu()
    opcion = input("Elija una opcion: ").strip()

    if opcion == "0":
      print("Saliendo del programa. ¡Hasta luego!")
      break
    elif opcion in opciones:
      opciones[opcion]()
    else:
      print("Opcion invalida, intente de nuevo.")


if __name__ == "__main__":
  main()
```

Escriba un programa en python que determine si es una palabra introducida por el usuario es palindrome

```python
def es_palindromo(palabra):
  #Determina si una palabra es palindromo
  normalizada = palabra.lower().replace(" ", "")
  return normalizada == normalizada[::-1]


def main():
  palabra = input("Ingrese una palabra: ").strip()
  if es_palindromo(palabra):
    print(f"\n'{palabra}' SI es un palindromo.")
  else:
    print(f"\n'{palabra}' NO es un palindromo.")


if __name__ == "__main__":
  main()
```

```python
from functools import reduce
import json


class SistemaEstadistico:

  def __init__(self, serie):
    # Convertimos todo a float
    self.serie = [float(x) for x in serie]

  def media(self):
    # Media aritmetica: suma de todos los valores entre la cantidad de valores
    suma = reduce(lambda acum, num: acum + num, self.serie, 0)
    return suma / len(self.serie)

  def mediana(self):
    # Mediana: el valor que queda justo en medio cuando los datos estan ordenados
    datos_ordenados = sorted(self.serie)
    n = len(datos_ordenados)
    mitad = n // 2
    if n % 2 == 0:
      return (datos_ordenados[mitad - 1] + datos_ordenados[mitad]) / 2
    else:
      return datos_ordenados[mitad]

  def frecuencia(self):
    # Frecuencia: cuantas veces se repite cada valor dentro de la serie
    frec = {}
    for valor in self.serie:
      frec[valor] = frec.get(valor, 0) + 1
    return frec

  def moda(self):
    # Moda: el valor (o valores) que mas se repite
    frec = self.frecuencia()
    frecuencia_maxima = max(frec.values())
    if frecuencia_maxima == 1:
      return None  # si nadie se repite, no hay moda
    modas = [valor for valor, cant in frec.items() if cant == frecuencia_maxima]
    return modas

  def varianza(self):
    # Varianza: que tanto se alejan en promedio los datos de la media (al cuadrado)
    media = self.media()
    suma = reduce(lambda acum, num: acum + (num - media) ** 2, self.serie, 0)
    return suma / len(self.serie)

  def desviacion_estandar(self):
    # Desviacion estandar: raiz cuadrada de la varianza
    return self.varianza() ** 0.5

  def desviacion_tipica(self):
    # Desviacion tipica es el mismo calculo que la desviacion estandar
    # (raiz cuadrada de la varianza); solo cambia el nombre segun la region.
    return self.varianza() ** 0.5

  def rango(self):
    # Rango: la diferencia entre el valor mas grande y el mas pequeno
    return max(self.serie) - min(self.serie)

  def resumen(self):
    print("\n ESTADISTICAS")
    print("Serie: ", self.serie)
    print("Media aritmetica: ", round(self.media(), 2))
    print("Mediana: ", round(self.mediana(), 2))
    moda = self.moda()
    print("Moda: ", moda if moda else "No hay moda (todos los valores se repiten igual)")
    print("Varianza: ", round(self.varianza(), 2))
    print("Desviacion estandar: ", round(self.desviacion_estandar(), 2))
    print("Desviacion tipica: ", round(self.desviacion_tipica(), 2))
    print("Rango: ", round(self.rango(), 2))
    print("Frecuencia: ", self.frecuencia())



serie_usuario = input("Ingrese una serie de numeros separados por coma: ").split(",")
sistemaEstadistico = SistemaEstadistico(serie_usuario)
sistemaEstadistico.resumen()
```

```python
from functools import reduce
import json

class Estudiante:
  """Representa a un estudiante con sus datos academicos y personales."""

  def __init__(self, nombre, materia, notas, edad, tiempo_traslado, horas_telefono):
    self.nombre = nombre
    self.materia = materia
    self.notas = notas                      # lista de notas del estudiante
    self.edad = edad
    self.tiempo_traslado = tiempo_traslado  # minutos para llegar a la UGB
    self.horas_telefono = horas_telefono    # horas diarias de uso del telefono

  def promedio_notas(self):
    return reduce(lambda acum, n: acum + n, self.notas, 0) / len(self.notas)

  def aprobado(self):
    return self.promedio_notas() >= 6

  def to_dict(self):
    return {
        "nombre": self.nombre,
        "materia": self.materia,
        "notas": self.notas,
        "edad": self.edad,
        "tiempo_traslado_min": self.tiempo_traslado,
        "horas_telefono": self.horas_telefono,
        "promedio": round(self.promedio_notas(), 2)
    }


class SistemaAcademico:
  """Administra una lista de estudiantes y calcula estadisticas del grupo."""

  def __init__(self):
    self.estudiantes = []

  def agregar_estudiante(self, estudiante):
    self.estudiantes.append(estudiante)

  def listar_estudiantes(self):
    for est in self.estudiantes:
      print(est.to_dict())

  def promedio_general(self):
    promedios = list(map(lambda est: est.promedio_notas(), self.estudiantes))
    return reduce(lambda acum, p: acum + p, promedios, 0) / len(promedios)

  def edad_promedio(self):
    edades = list(map(lambda est: est.edad, self.estudiantes))
    return reduce(lambda acum, e: acum + e, edades, 0) / len(edades)

  def tiempo_traslado_promedio(self):
    tiempos = list(map(lambda est: est.tiempo_traslado, self.estudiantes))
    return reduce(lambda acum, t: acum + t, tiempos, 0) / len(tiempos)

  def horas_telefono_promedio(self):
    horas = list(map(lambda est: est.horas_telefono, self.estudiantes))
    return reduce(lambda acum, h: acum + h, horas, 0) / len(horas)

  def aprobados(self):
    return list(filter(lambda est: est.aprobado(), self.estudiantes))

  def reprobados(self):
    return list(filter(lambda est: not est.aprobado(), self.estudiantes))

  def estudiantes_por_materia(self, materia):
    return list(filter(lambda est: est.materia.lower() == materia.lower(), self.estudiantes))

  def a_json(self):
    return json.dumps([est.to_dict() for est in self.estudiantes], indent=2, ensure_ascii=False)


class SistemaEstadistico:
  """Calcula estadisticas basicas (min, max, promedio) de una lista de valores."""

  def __init__(self, valores):
    self.valores = valores

  def minimo(self):
    return min(self.valores)

  def maximo(self):
    return max(self.valores)

  def promedio(self):
    return reduce(lambda acum, v: acum + v, self.valores, 0) / len(self.valores)

  def resumen(self):
    print(f"Minimo: {self.minimo()}")
    print(f"Maximo: {self.maximo()}")
    print(f"Promedio: {round(self.promedio(), 2)}")


# ----------------------------------------------------------------------
# Datos iniciales
# ----------------------------------------------------------------------

datos_estudiantes = [
    ("Francisco Saravia",   "Programacion III", [8, 7, 9],  20, 35, 5),
    ("Maria Fernanda Lopez","Programacion III", [6, 5, 7],  19, 20, 3),
    ("Jose Carlos Hernandez","Base de Datos",    [9, 8, 10], 21, 45, 4),
    ("Ana Beatriz Ramirez", "Programacion III", [4, 5, 6],  18, 15, 6),
    ("Luis Fernando Chavez","Redes",             [7, 6, 8],  22, 60, 2),
    ("Karen Vanessa Rivas", "Programacion III", [9, 9, 8],  20, 25, 5),
    ("Diego Alejandro Ponce","Base de Datos",    [5, 4, 6],  23, 50, 7),
    ("Andrea Sofia Martinez","Programacion III", [8, 7, 7],  19, 30, 3),
    ("Carlos Manuel Guevara","Redes",             [6, 7, 5],  24, 40, 4),
    ("Gabriela Alexandra Cruz","Programacion III",[10, 9, 9], 20, 20, 2),
    ("Oscar Eduardo Portillo","Base de Datos",    [3, 4, 5],  21, 55, 8),
    ("Fatima Elizabeth Amaya","Programacion III", [7, 8, 6],  18, 10, 3),
    ("Ricardo Antonio Melgar","Redes",             [8, 8, 9],  22, 35, 5),
    ("Mirna Yesenia Alvarado","Programacion III", [5, 6, 4],  19, 45, 6),
    ("Jonathan David Escobar","Base de Datos",    [9, 7, 8],  23, 25, 4),
    ("Patricia Noemi Flores","Programacion III",  [6, 6, 7],  20, 30, 5),
    ("Wilfredo Alexander Sosa","Redes",            [4, 5, 3],  25, 65, 7),
    ("Cindy Marisol Aguilar","Programacion III",  [10, 10, 9],21, 15, 2),
    ("Henry Alberto Martir", "Base de Datos",     [7, 6, 8],  20, 40, 4),
    ("Sandra Lisbeth Chacon","Programacion III",  [8, 9, 7],  19, 20, 3),
]


def construir_sistema():
  sistema = SistemaAcademico()
  for nombre, materia, notas, edad, traslado, horas_tel in datos_estudiantes:
    sistema.agregar_estudiante(
        Estudiante(nombre, materia, notas, edad, traslado, horas_tel)
    )
  return sistema


# ----------------------------------------------------------------------
# Funciones del menu
# ----------------------------------------------------------------------

def mostrar_listado_json(sistema):
  print("\n--- LISTADO DE ESTUDIANTES (JSON) ---")
  print(sistema.a_json())


def mostrar_estadisticas_grupo(sistema):
  print("\n--- ESTADISTICAS DEL GRUPO ---")
  print("Promedio general de notas: ", round(sistema.promedio_general(), 2))
  print("Edad promedio: ", round(sistema.edad_promedio(), 2))
  print("Tiempo de traslado promedio (min): ", round(sistema.tiempo_traslado_promedio(), 2))
  print("Horas de uso del telefono promedio: ", round(sistema.horas_telefono_promedio(), 2))


def mostrar_aprobados_reprobados(sistema):
  aprobados = sistema.aprobados()
  reprobados = sistema.reprobados()
  print("\nCantidad de aprobados: ", len(aprobados))
  print("Cantidad de reprobados: ", len(reprobados))

  print("\nNombres de los aprobados:")
  for est in aprobados:
    print(" -", est.nombre)

  print("\nNombres de los reprobados:")
  for est in reprobados:
    print(" -", est.nombre)


def mostrar_clasificacion_telefono(sistema):
  print("\n--- CLASIFICACION POR HORAS DE USO DEL TELEFONO ---")
  for est in sistema.estudiantes:
    if est.horas_telefono <= 2:
      categoria = "Uso bajo"
    elif est.horas_telefono <= 5:
      categoria = "Uso moderado"
    else:
      categoria = "Uso alto"
    print(f"{est.nombre}: {est.horas_telefono} hrs -> {categoria}")


def mostrar_estudiantes_por_materia(sistema):
  materia_buscada = input("Ingrese el nombre de la materia a buscar: ").strip()
  estudiantes_materia = sistema.estudiantes_por_materia(materia_buscada)
  print(f"\nEstudiantes que llevan {materia_buscada}: {len(estudiantes_materia)}")
  for est in estudiantes_materia:
    print(" -", est.nombre)


def mostrar_analisis_edades(sistema):
  edades = [est.edad for est in sistema.estudiantes]
  print("\n--- ANALISIS ESTADISTICO DE LAS EDADES DEL GRUPO ---")
  SistemaEstadistico(edades).resumen()


def mostrar_analisis_traslados(sistema):
  traslados = [est.tiempo_traslado for est in sistema.estudiantes]
  print("\n--- ANALISIS ESTADISTICO DEL TIEMPO DE TRASLADO A LA UGB ---")
  SistemaEstadistico(traslados).resumen()


# ----------------------------------------------------------------------
# Menu principal
# ----------------------------------------------------------------------

def mostrar_menu():
  print("\n===== SISTEMA ACADEMICO - MENU =====")
  print("1. Ver listado de estudiantes (JSON)")
  print("2. Ver estadisticas del grupo")
  print("3. Ver aprobados y reprobados")
  print("4. Ver clasificacion por uso del telefono")
  print("5. Buscar estudiantes por materia")
  print("6. Ver analisis estadistico de edades")
  print("7. Ver analisis estadistico de tiempos de traslado")
  print("0. Salir")


def main():
  sistema = construir_sistema()

  opciones = {
      "1": mostrar_listado_json,
      "2": mostrar_estadisticas_grupo,
      "3": mostrar_aprobados_reprobados,
      "4": mostrar_clasificacion_telefono,
      "5": mostrar_estudiantes_por_materia,
      "6": mostrar_analisis_edades,
      "7": mostrar_analisis_traslados,
  }

  while True:
    mostrar_menu()
    opcion = input("Elija una opcion: ").strip()

    if opcion == "0":
      print("Saliendo del sistema. ¡Hasta luego!")
      break
    elif opcion in opciones:
      opciones[opcion](sistema)
    else:
      print("Opcion invalida, intente de nuevo.")


if __name__ == "__main__":
  main()
```
