```python
#Librerias a Importar
import tensorflow as tf
import pandas as pd
import numpy as np
```

```python
#Leer archivo utilizando punto y coma como separador
datos = pd.read_csv('/content/conversor.csv', sep=';')
print(datos["farenheit"])
```

```python
#Obtener datos
c = datos["celsius"]
f = datos["farenheit"]
print(c,f)
```

```python
from ast import mod
#Crear Modelo
modelo = tf.keras.Sequential([
    tf.keras.layers.Dense(units=1, input_shape=[1])
])

```

```python
#compilar el modelo
modelo.compile(
    optimizer=tf.keras.optimizers.Adam(1.0),
    loss='mean_squared_error'
)
```

```python
#entrenar modelo
epocas = modelo.fit(c,f, epochs=200, verbose=1)
print("Entrenamiento Terminado")
```

```python
respuesta = modelo.predict(np.array([100.0]))
print(respuesta)
```
