```python
print("Sumar dos numeros")
num1 = float(input("Num 1:"))
num2 = float(input("Num 2:"))
respuesta = num1 + num2
print("La respuesta es", respuesta)
print("El tipo de dato de respuesta es", type(respuesta),"y el tipo de dato de num1 es", type(num1),"y el tipo de dato de num2 es", type(num2))
```
